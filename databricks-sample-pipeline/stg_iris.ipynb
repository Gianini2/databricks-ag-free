```python
CATALOG = "workspace"

df_raw = spark.table(f"{CATALOG}.raw.iris_dataframe")

display(df_raw.limit(10))  # visualização do Databricks
```

```python
stg_iris = df_raw.select(
    "sepal_length", "sepal_width", "petal_length", "petal_width", "species"
)

(stg_iris.write
    .mode("overwrite")
    .saveAsTable(f"{CATALOG}.staging.stg_iris"))
```
